```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%pip install matplotlib seaborn openpyxl

sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", None)

# Load dataset
df = pd.read_csv(
    r"C:\Users\Sanika Parab\Downloads\WA_Fn-UseC_-HR-Employee-Attrition.csv"
)

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
display(df.head())
```

```python
# Dataset overview
print("DATASET INFORMATION")
print("=" * 50)
print("Shape:", df.shape)
print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types and Non-Null Values:")
df.info()

print("\nStatistical Summary:")
display(df.describe(include="all"))
```

```python
# Data Dictionary
data_dictionary = pd.DataFrame({
    "Column Name": df.columns,
    "Data Type": df.dtypes.astype(str).values,
    "Missing Values": df.isnull().sum().values,
    "Unique Values": df.nunique().values
})

print("DATA DICTIONARY")
display(data_dictionary)

data_dictionary.to_excel("Data_Dictionary.xlsx", index=False)
print("Data dictionary saved as Data_Dictionary.xlsx")
```

```python
# Data quality checks and cleaning
print("MISSING VALUES")
missing = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Percentage": (df.isnull().sum() / len(df) * 100).round(2)
})
display(missing[missing["Missing Values"] > 0])

if df.isnull().sum().sum() == 0:
    print("No missing values found.")

print("\nDuplicate Rows:", df.duplicated().sum())

df_clean = df.drop_duplicates().copy()
df_clean["Attrition_Numeric"] = df_clean["Attrition"].map({"Yes": 1, "No": 0})

print("\nCleaned dataset shape:", df_clean.shape)
```

```python
# Numerical and categorical columns
numerical_columns = df_clean.select_dtypes(include=np.number).columns.tolist()
categorical_columns = df_clean.select_dtypes(include=["object", "string"]).columns.tolist()

print("NUMERICAL COLUMNS:")
print(numerical_columns)

print("\nCATEGORICAL COLUMNS:")
print(categorical_columns)
```

```python
# Key Performance Indicator (KPIs)
total_employees = len(df_clean)
employees_left = (df_clean["Attrition"] == "Yes").sum()
employees_stayed = (df_clean["Attrition"] == "No").sum()
attrition_rate = df_clean["Attrition_Numeric"].mean() * 100

kpi_summary = pd.DataFrame({
    "KPI": [
        "Total Employees",
        "Employees Left",
        "Employees Stayed",
        "Attrition Rate (%)",
        "Average Age",
        "Average Monthly Income",
        "Average Job Satisfaction"
    ],
    "Value": [
        total_employees,
        employees_left,
        employees_stayed,
        round(attrition_rate, 2),
        round(df_clean["Age"].mean(), 2),
        round(df_clean["MonthlyIncome"].mean(), 2),
        round(df_clean["JobSatisfaction"].mean(), 2)
    ]
})

print("KEY HR METRICS")
display(kpi_summary)
```

```python
# Attrition by Gender
gender_attrition = (
    df_clean.groupby("Gender")["Attrition_Numeric"].mean() * 100
).round(2)

print("ATTRITION RATE BY GENDER")
display(gender_attrition.to_frame("Attrition Rate (%)"))

plt.figure(figsize=(7, 5))
gender_attrition.plot(kind="bar")
plt.title("Attrition Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=0)
plt.show()

# Attrition by Department
department_attrition = (
    df_clean.groupby("Department")["Attrition_Numeric"].mean() * 100
).sort_values(ascending=False).round(2)

print("ATTRITION RATE BY DEPARTMENT")
display(department_attrition.to_frame("Attrition Rate (%)"))

plt.figure(figsize=(8, 5))
department_attrition.plot(kind="bar")
plt.title("Attrition Rate by Department")
plt.xlabel("Department")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=0)
plt.show()
```

```python
# Attrition by Job Role
jobrole_attrition = (
    df_clean.groupby("JobRole")["Attrition_Numeric"].mean() * 100
).sort_values(ascending=False).round(2)

print("ATTRITION RATE BY JOB ROLE")
display(jobrole_attrition.to_frame("Attrition Rate (%)"))

plt.figure(figsize=(11, 5))
jobrole_attrition.plot(kind="bar")
plt.title("Attrition Rate by Job Role")
plt.xlabel("Job Role")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=45, ha="right")
plt.show()

# Attrition by Age Group
df_clean["AgeGroup"] = pd.cut(
    df_clean["Age"],
    bins=[17, 25, 35, 45, 55, 65],
    labels=["18-25", "26-35", "36-45", "46-55", "56-65"]
)

age_attrition = (
    df_clean.groupby("AgeGroup", observed=True)["Attrition_Numeric"].mean() * 100
).round(2)

print("ATTRITION RATE BY AGE GROUP")
display(age_attrition.to_frame("Attrition Rate (%)"))

plt.figure(figsize=(8, 5))
age_attrition.plot(kind="bar")
plt.title("Attrition Rate by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=0)
plt.show()
```

```python
# Average Monthly Income by Attrition

# Average Monthly Income by Attrition

income_attrition = df_clean.groupby("Attrition")["MonthlyIncome"].mean()

display(income_attrition.to_frame("Average Monthly Income"))

income_attrition = df_clean.groupby("Attrition")["MonthlyIncome"].mean()

income_attrition.plot(kind="bar", figsize=(7, 5))

plt.title("Average Monthly Income by Attrition")
plt.xlabel("Attrition")
plt.ylabel("Average Monthly Income")
plt.xticks(rotation=0)

plt.show()

# Attrition by Overtime
overtime_attrition = (
    df_clean.groupby("OverTime")["Attrition_Numeric"].mean() * 100
).round(2)

print("ATTRITION RATE BY OVERTIME")
display(overtime_attrition.to_frame("Attrition Rate (%)"))

plt.figure(figsize=(7, 5))
overtime_attrition.plot(kind="bar")
plt.title("Attrition Rate by Overtime")
plt.xlabel("Overtime")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=0)
plt.show()
```

```python
# Attrition by Job Satisfaction
job_satisfaction_attrition = (
    df_clean.groupby("JobSatisfaction")["Attrition_Numeric"].mean() * 100
).round(2)

print("ATTRITION RATE BY JOB SATISFACTION")
display(job_satisfaction_attrition.to_frame("Attrition Rate (%)"))

plt.figure(figsize=(8, 5))
job_satisfaction_attrition.plot(kind="bar")
plt.title("Attrition Rate by Job Satisfaction")
plt.xlabel("Job Satisfaction Level")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=0)
plt.show()

# Attrition by Work-Life Balance
worklife_attrition = (
    df_clean.groupby("WorkLifeBalance")["Attrition_Numeric"].mean() * 100
).round(2)

print("ATTRITION RATE BY WORK-LIFE BALANCE")
display(worklife_attrition.to_frame("Attrition Rate (%)"))

plt.figure(figsize=(8, 5))
worklife_attrition.plot(kind="bar")
plt.title("Attrition Rate by Work-Life Balance")
plt.xlabel("Work-Life Balance Level")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=0)
plt.show()
```

```python
# Additional Employee Segment Analysis

# Marital Status
marital_attrition = (
    df_clean.groupby("MaritalStatus")["Attrition_Numeric"].mean() * 100
).sort_values(ascending=False).round(2)

print("ATTRITION RATE BY MARITAL STATUS")
display(marital_attrition.to_frame("Attrition Rate (%)"))

plt.figure(figsize=(8, 5))
marital_attrition.plot(kind="bar")
plt.title("Attrition Rate by Marital Status")
plt.xlabel("Marital Status")
plt.ylabel("Attrition Rate (%)")
plt.xticks(rotation=0)
plt.show()

# Years at Company
years_attrition = (
    df_clean.groupby("YearsAtCompany")["Attrition_Numeric"].mean() * 100
).round(2)

print("ATTRITION RATE BY YEARS AT COMPANY")
display(years_attrition.to_frame("Attrition Rate (%)"))

plt.figure(figsize=(10, 5))
plt.plot(years_attrition.index, years_attrition.values, marker="o")
plt.title("Attrition Rate by Years at Company")
plt.xlabel("Years at Company")
plt.ylabel("Attrition Rate (%)")
plt.show()
```

```python
# Key Insights
highest_department = department_attrition.idxmax()
highest_role = jobrole_attrition.idxmax()
highest_age_group = age_attrition.idxmax()
highest_overtime = overtime_attrition.idxmax()

print("=" * 70)
print("KEY INSIGHTS")
print("=" * 70)
print(f"1. Overall attrition rate: {attrition_rate:.2f}%")
print(f"2. Employees who left: {employees_left}")
print(f"3. Employees who stayed: {employees_stayed}")
print(f"4. Highest department attrition: {highest_department} ({department_attrition.max():.2f}%)")
print(f"5. Highest job role attrition: {highest_role} ({jobrole_attrition.max():.2f}%)")
print(f"6. Highest age-group attrition: {highest_age_group} ({age_attrition.max():.2f}%)")
print(f"7. Highest overtime-group attrition: {highest_overtime} ({overtime_attrition.max():.2f}%)")

print("\nBUSINESS RECOMMENDATIONS")
print("=" * 70)
print("1. Reduce Excessive Overtime - Employees who work overtime have a much higher attrition rate. The company should review workloads, improve staffing, and distribute work more evenly to reduce excessive overtime.")
print("")
print("2. Focus on High-Risk Departments and Job Roles - The Sales department and Sales Representatives showed particularly high attrition rates. The company should investigate the reasons behind turnover in these areas and provide better support, incentives, and career development opportunities.")
print("")
print("3. Improve Employee Compensation - Employees in the low-income group had a higher attrition rate. The company should review salary structures and ensure that compensation remains competitive and fair.")
print("")
print("4. Improve Job Satisfaction - Employees with low job satisfaction were more likely to leave. Regular employee feedback surveys, recognition programs, and opportunities for growth can help improve job satisfaction.")
print("")
print("5. Support Younger Employees - Employees aged 18–25 had the highest attrition rate. The company could introduce mentorship programs, career development opportunities, and better onboarding to improve retention among younger employees.")
print("")
print("6. Improve Work-Life Balance - Poor work-life balance was associated with higher attrition. Flexible working arrangements, better workload management, and employee wellness initiatives may help employees maintain a healthier balance.")
print("")
print("7. Develop Targeted Retention Strategies - Instead of applying the same solution to every employee, the company should focus retention efforts on groups with higher attrition rates, such as employees working overtime, younger employees, low-income employees, and employees in high-attrition job roles.")
print("")
print("8. Conduct Regular Employee Feedback - Regular surveys and feedback sessions can help management identify employee concerns early and take action before employees decide to leave.")
print("")
print("Overall Recommendation - The company should prioritize reducing excessive overtime, improving work-life balance, increasing job satisfaction, and developing targeted retention strategies for high-risk employee groups. These actions could help reduce employee attrition and improve overall employee retention.")
print("=" * 70)
```
